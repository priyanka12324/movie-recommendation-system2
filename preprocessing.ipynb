```python
def convert(obj):
  l=[]
  for i in ast.literal_eval(obj):
      l.append(i['name'])
  return l    
```

```python
def convert3(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i["name"])
            counter+=1
        else:
            break
    return L        
```

```python
def fetch_director(obj):
  l=[]
  for i in ast.literal_eval(obj):
     if i['job']=='Director': 
        l.append(i['name'])
        break 
  return l    
```
